```python
with open("/data/igor/dedup_data/queries/fin","wb") as f:
    for text in [" on Tuesday", " on Monday"]:
        arr = np.array(tokenizer.encode(text), dtype=np.uint16).view(np.uint8).tobytes()
        length = np.array([len(arr)], dtype=np.uint32).view(np.uint32).tobytes()
        f.write(length)
        f.write(arr)
```
